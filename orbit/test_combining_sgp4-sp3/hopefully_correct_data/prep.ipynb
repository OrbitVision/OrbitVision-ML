```python
# Combine csv
import pandas as pd
```

```python
ds1 = pd.read_csv("sp3_C38_2024_onePerDay.csv")
ds2 = pd.read_csv("sp3_C38_2025_onePerDay.csv")

merged = pd.concat([ds1, ds2], ignore_index=True)
merged.to_csv("sp3_C38_2024-25.csv", index=False, encoding='utf-8')
```

```python
#Need to use sp3 csv to create final ds with sp3 reference
df_sgp4 = pd.read_csv("ECEF_C38_2024-25_4pd_long_format.csv")
df_sp3 = pd.read_csv("sp3_C38_2024-25.csv")

df_final = pd.merge(df_sgp4, df_sp3, left_on=["target_epoch", "sat_id"], right_on=["epoch", "sat_id"], how="inner")

df_final.to_csv("final.csv", index=False, encoding='utf-8')
```

```python
df = pd.read_csv("final.csv")

df = df.rename(columns={
    'x_x': 'x', 'y_x': 'y', 'z_x': 'z',
    'x_y': 'sp3_x', 'y_y': 'sp3_y', 'z_y': 'sp3_z'
})
df.to_csv("final.csv", index=False, encoding='utf-8')
```

```python
#Here there is need to modify the final ds to change sp3 data into errors
df = pd.read_csv("final.csv")
df['x_error'] = df['sp3_x'] - df['x']
df['y_error'] = df['sp3_y'] - df['y']
df['z_error'] = df['sp3_z'] - df['z']

columns_to_keep = [
    'tle_epoch', 'target_epoch', 'days_offset', 'bstar', 'x', 'y', 'z', 'vx', 'vy', 'vz', 'x_error', 'y_error', 'z_error'
    ]

df_final = df[columns_to_keep]

df_final.to_csv("C38_ready.csv", index=False, encoding='utf-8')
```
